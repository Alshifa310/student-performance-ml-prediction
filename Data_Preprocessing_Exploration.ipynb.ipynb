### Part 1 : Training Algorithem


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sp_df = pd.read_csv("StudentPerformanceFactors.csv")

sp_df.drop('Gender', axis=1, inplace=True)
sp_df.drop('Distance_from_Home', axis=1, inplace=True)
sp_df.drop('Family_Income', axis=1, inplace=True)
sp_df.drop('School_Type', axis=1, inplace=True)
sp_df.drop('Parental_Education_Level', axis=1, inplace=True)
sp_df.drop('Physical_Activity', axis=1, inplace=True)

sp_df.dropna(axis=0, inplace=True)

sp_df.drop(sp_df[sp_df['Exam_Score'] == 101].index, inplace=True)

sp_df.replace('Low',0, inplace=True)
sp_df.replace('Medium',1 , inplace=True)
sp_df.replace('High',2 , inplace=True)

sp_df.replace('No', 0, inplace=True)
sp_df.replace('Yes', 1, inplace=True)

sp_df.replace('Neutral', 0, inplace=True)
sp_df.replace('Positive', 1, inplace=True)
sp_df.replace('Negative', 2, inplace=True)

X, y = sp_df.drop('Exam_Score', axis=1), sp_df['Exam_Score'].values

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=87)

from sklearn.linear_model import LinearRegression
lm = LinearRegression()
lm.fit(X_train,y_train)
```

### Part 4 : Explaining 

```python
@app.get("/student/explain")
def explain():
    explanation = """
    1. Hours_Studied: The number of hours the student studied. This should be a numeric value indicating 
       the total hours spent on studying. Example: 23 (numeric).

    2. Attendance: The percentage of classes attended by the student. This should be a numeric value 
       from 0 to 100. Example: 84 (numeric).

    3. Parental_Involvement: The level of involvement of the student's parents. 
       Values: 0 = Low, 1 = Medium, 2 = High. Example: 0 (Low), 1 (Medium), 2 (High).

    4. Access_to_Resources: The availability of educational resources for the student. 
       Values: 0 = Low, 1 = Medium, 2 = High. Example: 2 (High).

    5. Extracurricular_Activities: Whether the student participates in extracurricular activities. 
       Values: 0 = No, 1 = Yes. Example: 0 (No), 1 (Yes).

    6. Sleep_Hours: The number of hours the student sleeps daily. This should be a numeric value. 
       Example: 7 (numeric).

    7. Previous_Scores: The student's previous exam scores. This should be a numeric value between 0 and 100. 
       Example: 73 (numeric).

    8. Motivation_Level: The student's motivation level. Values: 0 = Low, 1 = Medium, 2 = High. 
       Example: 0 (Low), 1 (Medium), 2 (High).

    9. Internet_Access: Whether the student has access to the internet. 
       Values: 0 = No, 1 = Yes. Example: 1 (Yes).

    10. Tutoring_Sessions: The number of tutoring sessions attended by the student. 
        This should be a numeric value. Example: 0 (numeric).

    11. Teacher_Quality: The quality of teaching received by the student. 
        Values: 0 = Low, 1 = Medium, 2 = High. Example: 1 (Medium), 2 (High).

    12. Peer_Influence: The influence of peers on the student. 
        Values: 0 = Neutral, 1 = Positive, 2 = Negative. Example: 0 (Neutral), 1 (Positive), 2 (Negative).

    13. Learning_Disabilities: Whether the student has any learning disabilities. 
        Values: 0 = No, 1 = Yes. Example: 0 (No), 1 (Yes).

    14. Exam_Score: This is the output prediction for how much the student will score in their next exam. 
        It is a numeric value predicted based on the provided features.
    """
    return explanation
```

### Part 5 : Evaluate

```python
@app.post("/student/evaluate")
def test_sample():
    if request.is_json:
        person = request.get_json()
        df = pd.DataFrame([person])
        result = lm.predict(df)
        return '{}'.format(result[0])
    return {"error": "Request must be JSON"}
```
