### Part 6 : Consuming Web Services

```python
import requests
```

#### 1. Checking Sample Request

```python
sample_response = requests.get("http://127.0.0.1:5000//student/sample")

import json

def jprint(obj):
    # create a formatted string of the Python JSON object
    text = json.dumps(obj, sort_keys=True, indent=4)
    print(text)

jprint(sample_response.json())
```

#### 2. Checking Explain Request

```python
explain_response = requests.get("http://127.0.0.1:5000//student/explain")

print(explain_response.text)
```
